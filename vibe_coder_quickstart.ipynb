### 2. Load Model in 4-bit (Fast & Zero Memory Overhead)

```python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig, TextStreamer

MODEL_ID = "shawaz03/vibe-coder-7b-max"

print(f"🚀 Loading {MODEL_ID} in 4-bit NF4 for Google Colab...")

# 4-bit configuration (Fits smoothly in 5.2 GB VRAM on Colab T4)
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_use_double_quant=True,
    bnb_4bit_compute_dtype=torch.float16,
)

tokenizer = AutoTokenizer.from_pretrained(MODEL_ID, trust_remote_code=True)
model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID,
    quantization_config=bnb_config,
    device_map="auto",
    trust_remote_code=True,
)
model.eval()
streamer = TextStreamer(tokenizer, skip_prompt=True, skip_special_tokens=True)

print("\n✅ VIBE CODER IS READY TO GENERATE CODE!")
```

### 3. Interactive Code Generator

```python
# Vibe Coder System Prompt
VIBE_CODER_SYSTEM_PROMPT = """You are Vibe Coder, a world-class principal full-stack software engineer and UI/UX designer.
ENGINEERING DIRECTIVES:
1. ZERO PLACEHOLDERS: Never output '// TODO' or incomplete stubs. Every component, hook, and route must be fully written.
2. MODERN DESIGN SYSTEM: Use Tailwind CSS with dark neutral palettes (bg-neutral-900, border-neutral-800), clean accents (cyan, emerald, violet), and Lucide icons.
3. TYPE SAFETY: Include full TypeScript interfaces and prop types."""

def ask_vibe_coder(prompt_text, temperature=0.2, max_tokens=1500):
    messages = [
        {"role": "system", "content": VIBE_CODER_SYSTEM_PROMPT},
        {"role": "user", "content": prompt_text}
    ]
    
    formatted_prompt = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    inputs = tokenizer(formatted_prompt, return_tensors="pt").to("cuda")
    
    print("=" * 75)
    print(f"🚀 PROMPT: {prompt_text}")
    print("=" * 75 + "\n")
    
    with torch.no_grad():
        model.generate(
            **inputs,
            streamer=streamer,
            max_new_tokens=max_tokens,
            temperature=temperature,
            top_p=0.95,
            repetition_penalty=1.05,
            do_sample=True,
            eos_token_id=tokenizer.eos_token_id,
            pad_token_id=tokenizer.pad_token_id,
        )
    print("\n" + "=" * 75)
```

### 4. Run Example Prompts

```python
# Example 1: Full-Stack Component
ask_vibe_coder("Build an interactive pricing card in React with Tailwind CSS, monthly/annual toggle, and checkmark feature list.")
```

```python
# Example 2: Next.js Server Action
ask_vibe_coder("Write a Next.js Server Action with Zod validation to update user settings with error handling.")
```
